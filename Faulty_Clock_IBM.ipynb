```python
import cv2
import matplotlib.pyplot as plt
quesn=cv2.imread("Q_4_Faulty_Clock_IBM.png")
# 2. Set the display size (width, height in inches)
plt.figure(figsize=(200, 200))
plt.imshow(quesn)
```

```python
def FaultyClock(time:str)->str:
    parts=time.split(":")
    part_idx=-1
    ch_idx=-1
    for i in range(3):
        if " " in parts[i]:
            part_idx=i
            ch_idx=parts[i].index(" ")
            break
    target=parts[part_idx]
    if part_idx==0:
        if ch_idx==0:
            unit=int(target[1])
            max_tens=2 if unit<=3 else 1
            possible_values=list(v for v in range(0,max_tens+1))
        else:
            tens = int(target[0])
            max_unit = 3 if tens == 2 else 9
            possible_values = list(range(0, max_unit + 1))   

    elif part_idx == 1 or part_idx == 2:
        # Minutes or Seconds: valid values range from 00 to 59
        if ch_idx == 0:
            # Missing tens digit (0 to 5)
            possible_values = list(range(0, 6))
        else:
            # Missing units digit (0 to 9)
            possible_values = list(range(0, 10))

    min_d=str(possible_values[0])    
    max_d=str(possible_values[-1])
    min_parts = list(parts)
    max_parts = list(parts)
    min_parts[part_idx]=target.replace(" ",min_d)
    max_parts[part_idx]=target.replace(" ",max_d)
    min_time = ":".join(min_parts)
    max_time = ":".join(max_parts)
    print(f"Min Time:{min_time}")
    print(f"Max Time:{max_time}")

time=input().strip()
print(FaultyClock(time))
```

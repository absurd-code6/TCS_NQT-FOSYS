```python
def Pairwise_Difference(nums:list[int])->int:
    if not nums:
        return -1
    n=len(nums)
    sum=0
    for i in range(1,n,2):
        sum+=abs(nums[i]-nums[i-1])
    while sum>=10:
        dsum=0
        while sum!=0:
            dsum+=sum%10
            sum//=10
        sum=dsum
    return sum

print(Pairwise_Difference([15,5,20,8,50,10]))

        



    
```

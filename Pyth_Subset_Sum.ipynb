```python
def subset_sum(arr, h, tempsum, idx):
    # Base Case 1: If we reached the end of the array, check if we found the sum
    if idx >= len(arr):
        if h == tempsum:
            return 1
        else:
            return 0

    # Choice 1: Include the current element in the sum
    c1 = subset_sum(arr, h, tempsum + arr[idx], idx + 1)
    
    # Choice 2: Exclude the current element from the sum
    c2 = subset_sum(arr, h, tempsum, idx + 1)

    # Return the sum of both choices (modulo 10^9 + 7 to handle large results)
    return (c1 + c2) % 1000000007

# --- How to run a test case ---
'''if __name__ == "__main__":
    # Testcase 1 from your image
    arr = [2, 3, 5, 6, 8, 10]
    target_sum = 10
    
    # Start at tempsum = 0 and idx = 0
    result = subset_sum(arr, target_sum, 0, 0)
    print(result)  # Output: 3'''

t=int(input())
while t>0:
    n=int(input())
    arr=list(map(int,input().strip().split()))[:n]
    sum=int(input())
result = subset_sum(arr, target_sum, 0, 0)
print(result)
```

```python
def Floyd(n):
    if not n:
        return -1
    numbers=1
    for i in range(1,n+1):
        print("")
        for j in range(1,i+1):
            print(numbers,end=" ")
            numbers+=1
        print()    

Floyd(5)
Floyd(3)
```

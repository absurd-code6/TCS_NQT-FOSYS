```python
class BinaryTree:
    def __init__(self,value=0,left=None,right=None):
        self.value=value
        self.left=left
        self.right=right
    
def minDepth(root_node):
        # Scenario 1: The tree/node is empty
        if root_node is None:
            return 0
        # Scenario 2: The node is a leaf node (no children)
        if root_node.left is None and root_node.right is None:
            return 1
        # Scenario 3: If left child is missing, force search down the right side
        if root_node.left is None:
            return 1 + minDepth(root_node.right)
        # Scenario 4: If right child is missing, force search down the left side
        if root_node.right is None:
            return 1 + minDepth(root_node.left)
        # Scenario 5: Both children exist, pick the smaller depth
        return 1 + min(minDepth(root_node.left),minDepth(root_node.right))

# 3. Create the tree and test it
# We build the tree from the bottom up so we can link them easily.

# Step A: Create the bottom-most leaf node
node4 = BinaryTree(4)

# Step B: Create nodes 2 and 3. 
# Node 2 connects to Node 4 on its left. Node 3 has no children.
node2 = BinaryTree(2, left=node4)
node3 = BinaryTree(3)

# Step C: Create the root node (Node 1) and connect it to Node 2 and Node 3
root = BinaryTree(1, left=node2, right=node3)

# 4. Run the code and print the result
result = minDepth(root)
print("The minimum depth of the tree is:", result)
```

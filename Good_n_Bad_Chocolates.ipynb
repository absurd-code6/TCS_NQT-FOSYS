```python
def Good_n_Bad_Chocolates(s,c):
    requirements=list(map(int,s.strip().split()))
    n=len(c)
    satisfied=0
    i=0
    for req in requirements:
        while i<n and c[i]=='R':
            i+=1
        count=0
        j=i
        while j<n and c[j]=='G':
            count+=1
            j+=1
        if count>=req:
            satisfied+=1
            i+=req
        else:
            break
    return satisfied

```

```python
s="1 1 2 3"
c="RGRGGGGRGGR" 
print(Good_n_Bad_Chocolates(s,c))
```

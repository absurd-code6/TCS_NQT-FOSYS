```python
#Two Sum Problem
'''Imagine you are given a list of numbers and a target number. 
Your job is to find two different numbers inside that list that can be 
added together to equal the target number.

Once you find those two numbers, you don't return the numbers themselves; 
instead, you must return their positions (indices) in the list.

The Rules:
Exactly One Solution: You are guaranteed that there is always a single, 
perfect pair of numbers in the list that will work. 
You don't have to worry about what happens if there are zero solutions 
or multiple solutions.

No Double-Dipping: You cannot use the exact same 
element twice to get your answer. For example, if your 
target is 6 and there is a 3 in the list, you cannot just add 
that 3 to itself—you would need to find a second 3 at a 
different position in the list.

Order Doesn't Matter: You can list the two positions in 
any order you like (e.g., [1, 2] or [2, 1] are both perfectly correct).'''
```

```python
def Two_sum(arr,target):
    for i in range(len(arr)):
        for j in range(i+1,len(arr)):
            if arr[i]+arr[j]==target:
                return[i,j]

arr=list(map(int,input().split()))#[3,2,4]
target=int(input())#6
print(Two_sum(arr,target))#[1,2] i.e. 2+4=6
#TC=O(n^2)
```

```python
def Two_Sum(arr,target):
    seen={}
    for idx,num in enumerate(arr):
        #Calculate what number we need to find to hit the target
        complement=target - num
        #If the complement is already in our dictionary, we found our pair!
        if complement in seen:
            return [seen[complement],idx]
        
        # Otherwise, save the current number and its index to the dictionary
        seen[num] = idx
#TC=O(n)
        
```

# How it Works (Using Example 2: nums = [3, 2, 4], target = 6)                                                                                                                                                                             # First Iteration (index 0, num = 3):Complement = 6 - 3 = 3.Is 3 in seen? No.
Update dictionary: seen = {3: 0}


Second Iteration (index 1, num = 2):


Complement = 6 - 2 = 4$Is 4 in seen? No.


Update dictionary: seen = {3: 0, 2: 1}


Third Iteration (index 2, num = 4):Complement = $6 - 4 = 2$Is 2 in seen? Yes!


It's at index 1.The function immediately returns [1, 2]

# But how is the 3 removed or ignored?

It isn't actually removed or deleted from the dictionary—it is simply ignored because it fails the if condition.

Think of the if statement like a strict bouncer at a club door. The bouncer is only looking for one specific guest: the complement (2).

When the loop is at index 2 (num = 4), here is exactly what the seen dictionary looks like:

Python
seen = {

    3: 0,  # The number 3 is at index 0
    2: 1   # The number 2 is at index 1

}

Now, let's step through the code line-by-line for that iteration:

complement = target - num → 6 - 4 → 2

if complement in seen: → This asks Python: "Is the key 2 inside the dictionary?"

Python looks at the dictionary. It sees two keys: 3 and 2.

It looks at 3: Not a match. (Ignored!)

It looks at 2: Match found!

Because the condition if 2 in seen is True, Python enters the if block and executes:

Python
return [seen[2], index]  # seen[2] gives us index 1, and the current index is 2
The pair 3: 0 is still sitting safely inside the dictionary, but because 3 is not equal to 2, the code completely bypasses it. It never gets looked up, and it never gets returned.

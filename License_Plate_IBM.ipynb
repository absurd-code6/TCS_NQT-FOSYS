```python
lp=input().strip().upper()
if len(lp)<2 or len(lp)>10:
    print("Invalid")
    exit(0)
for char in lp:
    is_letter = 65 <= ord(char) <= 90   # 'A' - 'Z'
    is_digit = 48 <= ord(char) <= 57    # '0' - '9'
    
    # If it is NEITHER a letter NOR a digit
    if not (is_letter or is_digit):
        print("Invalid")
        break
else:
    print("Valid")

```

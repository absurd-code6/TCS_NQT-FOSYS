```python
import math
def getFare(source,destination):
    path = [800, 600, 750, 900, 1400, 1200, 1100, 1500]

    busStops = ["TH", "GA", "IC", "HA", "TE", "LU", "NI", "CA"]
    if source  not in busStops or destination not in busStops:
        return "INVALID INPUT"
    #We need the index numbers.
    s = busStops.index(source)
    d = busStops.index(destination)
    i=0
    distance=0
    while i!=d:
        distance+=path[i]
        i=(i+1)%len(busStops) # Since bus moves in circular fashion
    fare=math.ceil(distance/200)
    return fare

source=input().upper()
destination=input.upper()
print(getFare(source,destination))

    
```

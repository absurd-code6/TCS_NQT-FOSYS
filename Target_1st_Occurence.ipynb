```python
def Ist_occurence(arr,target):
    result=[]
    for i in range(len(arr)):
        if arr[i]==target:
            result.append(i)
    return result[0]
    #if the target value is not present in the array 
    if not result: # or if len(result)==0:
        return -1
```

```python
def arr_2_BT(arr)->TreeNode:
    def converter(l,r):
        if l>r:
            return 
        mid=(l+r)//2
        root=TreeNode(arr[mid])
        root.left=converter(l,mid-1)
        root.right=converter(mid+1,r)
        return root
    return converter(0,len(arr)-1)

#TC=O(n), SC=O(logn=ht of the tree)
```

```python
class TreeNode:
    def __init__(self,val=0,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right

```

```python
print(arr_2_BT([-10,-3,0,5,9]))
#The output in Python will be the string representation of the root 
# TreeNode object memory address
```

```python
from collections import deque
def print_lvlOrder(fx):
    if not fx:
        return
    res=[]
    queue=deque([fx])
    while queue:
        node=queue.popleft()
        if  node:
            res.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
        else:
            res.append(None)
    while res and res[-1] is None:
        res.pop()
    return res    

print(print_lvlOrder(arr_2_BT([-10,-3,0,5,9])))
```

```python
"""Key Concepts Behind the Function
Queue (FIFO - First In, First Out):

Think of a line at a grocery store checkout. The first person to stand in 
line is the first person served.

In code, queue = deque([root]) initializes the line with the top node. 
queue.popleft() takes the node at the front of the line.

Tracking None (Empty Branches):

Whenever a node has no left or right child, a None is added to the queue 
so we know a branch was missing.

Trimming Trailing Nones:

At the very end of the tree, the last level adds a bunch of empty 
None placeholders. The final while loop cleans those up from the
 end of the list so the output stays tidy."""
```

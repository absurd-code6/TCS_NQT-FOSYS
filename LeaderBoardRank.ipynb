# Proxy for Kth Largest Element

```python
import heapq
def LBRank(arr:list[int],k:int)->int:
    max_heap=[-x for x in arr]
    heapq.heapify(max_heap)
    #return -max_heap
    for i in range(k-1):
        -heapq.heappop(max_heap)
    kth_largest=-heapq.heappop(max_heap)
    return kth_largest

#TC=O(n+klogn), SC=O(n)

n=int(input())
arr=[int(input()) for _ in range(n)] #[20,70,50,90,30,40]
k=int(input("Enter the value of k:"))
print(LBRank(arr,k))
```

```python
def LBRank2(arr:list[int],k:int)->int:
    min_heap=arr[:k]
    heapq.heapify(min_heap)
    for i in arr[:k]:
        if i>min_heap[0]:
            heapq.heappushpop(min_heap,i)
    return min_heap[0]

#TC=O(nlogk),SC=O(1)

n=int(input())
arr=[int(input()) for _ in range(n)] #[20,70,50,90,30,40]
k=int(input("Enter the value of k:"))
print(LBRank(arr,k))
```

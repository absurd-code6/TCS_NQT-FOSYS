# This is a classic Sliding Window Problem

Correct Approach

Try every possible split:

Take 0 cards from the left and k from the right.


Take 1 from the left and k-1 from the right.


Take 2 from the left and k-2 from the right.


...
Take k from the left and 0 from the right

the most intuitive solution uses two pointers (left and right). 


The trick is that you don't move them independently like in a typical two-pointer problem. 

Instead, you gradually replace cards taken from the left with cards taken from the right

```python
def Max_Points(cardScore,k):
    curr_sum=sum(cardScore[:k])
    max_sum=curr_sum
    n=len(cardScore)
    left=k-1
    right=n-1
    while left>=0:
        curr_sum= (curr_sum-cardScore[left]) + cardScore[right]
        max_sum=max(max_sum,curr_sum)
        left-=1
        right-=1
    print(max_sum)        
        
```

Let’s strip away the backstory about the virus. The core of the question is:If there are

 N people in a room, and every person shakes hands with every other person exactly once,
 
  how many total handshakes happen?Let’s test this with small numbers to see if we can 
  
  find a pattern:If there is 1 person (N = 1): They have no one else to shake hands with.Total Handshakes = 0If there are 2 people (N = 2, let's call them A and B): A shakes 
  
  
hands with B. That's it.Total Handshakes = 1If there are 3 people (N = 3, let's call them 

A, B, and C): * A shakes hands with B and C (2 handshakes).B already shook hands with A,

 so B only needs to shake hands with C (1 handshake).C has already shaken hands with everyone.Total Handshakes = 2 + 1 = 3If there are 4 people (N = 4, A, B, C, D): * A 

 shakes hands with 3 people (B, C, D).B shakes hands with 2 new people (C, D).C shakes
 
  hands with 1 new person (D).Total Handshakes = 3 + 2 + 1 = 6💡 Finding the Pattern (The Formula)Notice what is happening. For N people, the total number of handshakes is the 
  
  sum of all numbers from 1 up to N - 1.For N = 4, the answer is 3 + 2 + 1 = 6.In 
  
  mathematics, there is a very famous, simple formula to calculate the sum of numbers 
  
  from 1 to x:\text{Sum} = \frac{x \times (x + 1)}{2}Since our x here is (N - 1), we can 
  
  substitute it into the formula:\text{Handshakes} = \frac{(N - 1) \times ((N - 1) + 1)}
  
  {2} = \frac{(N - 1) \times N}{2}So, our magic formula for N people is:\text{Handshakes}
   = \frac{N \times (N - 1)}{2}

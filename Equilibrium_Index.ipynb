# The Main Idea

Imagine a see-saw. We want to find a position (an index) where the numbers on the left side balance out (equal) the numbers on the right side. The number standing at the index itself doesn't belong to either side.Instead of recalculating the sums repeatedly, we can do this in two simple steps:

Find the total sum of all numbers in the array.Walk through the array from left to right, keeping track of the left sum.

Since Total Sum = Left Sum + Current Element + Right Sum, we can easily find the Right Sum as:

Right Sum = Total Sum -Left Sum - Current Element 

If Left Sum == Right Sum, we found our answer!

```python
def equilibrium_idx(arr):
    n=len(arr)
    total_sum=sum(arr)
    left_sum=0
    for i in range(n):
        right_sum=total_sum-left_sum-arr[i]
        if left_sum==right_sum:
            return i
        left_sum+=arr[i]
    return -1

arr=list(map(int,input().split(',')))# [-7,1,5,2,-4,3,0]
print(equilibrium_idx(arr))# 3
```

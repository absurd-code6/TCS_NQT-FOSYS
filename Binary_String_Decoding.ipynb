```python
def Binary_Str_Decoder(s:str)->str:
    if not s:
        return "-1"
    res=""
    count=0
    for c in s:
        if c=="1":
           count+=1
        else:
           if count > 0:
              res += chr(64 + count)
              count = 0
    if count > 0:
           res += chr(64 + count)
    return res if res else "-1"
```

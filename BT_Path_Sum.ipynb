```python
def BT_Path_Sum(root:Tree_node,t:int):
    if not root:
        return False
    def dfs(root,cursum):
        #cursum=0
        cursum+=root.val
        if not root.left and not root.right:
            return cursum==t
        return (dfs(root.left,cursum) or dfs(root.right,cursum))
    return dfs(root,0)
    
```

Case                         Time Complexity        Space Complexity  

Best / Average (Balanced Tree)   O(N)                     O(log N)

Worst Case (Skewed Tree)         O(N)                       O(N)

```python
class Tree_node:
    def __init__(self,val=0,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right

root=Tree_node(1)
```

```python
class TreeNode:
    def __init__(self,val=0,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right
root = TreeNode(1)
root.left = TreeNode(2, TreeNode(3), TreeNode(4))
root.right = TreeNode(2, TreeNode(4), TreeNode(3))
```

Let's break down this problem step-by-step.We need to explore a binary tree from the top (root) all the way down to the bottom nodes (leaves), adding up the values along the way. Among all paths that lead to a leaf, we want the one that gives us the largest sum, with one catch: that sum cannot be divisible by $K$.

Since N can be as large as 100,000, a simple recursive solution might crash Python due to a stack overflow (too many nested function calls). Instead, we will construct an iterative Depth-First Search (DFS) using a loop and a list (stack) to explore the paths safely.

```python
def dfs(node, parent, curr_sum):
    global max_sum
    curr_sum += values[node-1]
    is_leaf = True
    
    for nbr in adj[node]:
        if nbr != parent:
            is_leaf = False
            dfs(nbr, node, curr_sum)   
            
    # Fixed: This block must be outside the for-loop so it only checks AFTER looking at all neighbors
    if is_leaf and curr_sum % K != 0:
        max_sum = max(max_sum, curr_sum)

# --- Main Program Execution ---
N = int(input())
values = list(map(int, input().split()))

adj = {i: [] for i in range(1, N+1)}
for _ in range(N-1):
    u, v = map(int, input().split())
    adj[u].append(v)
    adj[v].append(u)
    
K = int(input())
max_sum = -1
# Start the DFS traversal
dfs(1, -1, 0)
print(max_sum)# o/p=21
'''Input Format:
7 #No ofnodes(=N) 
3 4 8 2 1 6 10 # N Nodes
1 2 # N-1 edges
1 3
2 4
2 5
3 6
3 7
5 # value of K'''
```

# Step 1: Building the Binary Tree (adj)
The program reads $N=7$ and stores the node weights in the values list. Because list indexing in Python starts at 0, Node 1's weight is at values[0] (3), Node 2 is at values[1] (4), and so on.

Next, the code loops $N-1$ times (6 times) to read the edges and construct the graph. It creates a dictionary called adj (short for adjacency list). Because a tree edge goes both ways initially, every pair u, v gets added to both nodes' lists:

# After reading all 6 edges, your adj dictionary looks exactly like this:
adj = {

1: [2, 3],       # Node 1 is connected to 2 and 3

2: [1, 4, 5],    # Node 2 is connected back to 1, and down to 4 and 5

3: [1, 6, 7],    # Node 3 is connected back to 1, and down to 6 and 7

4: [2],          # Node 4 is connected back to 2

5: [2],          # Node 5 is connected back to 2

6: [3],          # Node 6 is connected back to 3
    
7: [3]           # Node 7 is connected back to 3

}

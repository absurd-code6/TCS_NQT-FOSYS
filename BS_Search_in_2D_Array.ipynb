```python
def BS_Search2D(arr2d,target):
    if not arr2d:
        return -1
    m=len(arr2d) # m is the number of rows
    n=len(arr2d[0]) # n is the number of columns
    l=0
    h=(n*m)-1
    while l<=h:
        mid=(l+h)//2
        row=mid//n
        col=mid%n
        if arr2d[row][col]==target:
            return True
        elif arr2d[row][col]<target:
            l=mid+1
        else:
            h=mid-1
    return False

#TC=O(log2(m*n)),SC=O(1)
```

Search In A 2D Array/Matrix II

```python
import cv2
import matplotlib.pyplot as plt
quesn1=cv2.imread("Q_4_BS_Search_in_2D_Array_II.png")
# 2. Set the display size (width, height in inches)
plt.figure(figsize=(200, 200))
plt.imshow(quesn1)
```

```python
def BS_2D_ArrayII(arr2d1,t):
    if not arr2d1:
        return {-1,-1}
    m=len(arr2d1)
    n=len(arr2d1[0])
    row=0
    col=n-1
    while row<m and col>=0:
        if arr2d1[row][col]==t:
            return [row,col]
        elif arr2d1[row][col]<t:
            row+=1
        else:
            col-=1
    return [-1,-1]
        
#TC =O(m+n),SC = O(1)
```

```python
arr2d = [
    [1,  3,  5,  7],
    [10, 11, 16, 20],
    [23, 30, 34, 60]
]

target=int(input())
print(target)
print(BS_Search2D(arr2d,target))
```

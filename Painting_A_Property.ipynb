```python
interior_walls=int(input())
exterior_walls=int(input())
interior_cost=0
exterior_cost=0
for i in range(interior_walls):
    area=float(input())
    interior_cost+=area*18
    
for i in range(exterior_walls):
    area = float(input())
    exterior_cost = exterior_cost + area * 12

total_estimated_cost=interior_cost+exterior_cost
print(total_estimated_cost)
```

# Understand the problem

Painting costs:

Interior wall = ₹18 per sq.ft
Exterior wall = ₹12 per sq.ft

We have to take input:

Number of interior walls
Number of exterior walls
Area of every interior wall
Area of every exterior wall

Finally,

Total Cost = Interior Cost + Exterior Cost

Think about the logic

Suppose the user enters

Interior walls = 3
Exterior walls = 2

Now we need to ask for

Area of Interior wall 1
Area of Interior wall 2
Area of Interior wall 3

Area of Exterior wall 1
Area of Exterior wall 2

So we need a loop, because we don't know how many walls the user will enter.

# What if number of walls is 0?

Question says

If the number of walls is zero, skip taking the surface area.

Actually, the for loop automatically handles this.

Example

interior_walls = 0

Then

for i in range(0):

runs 0 times, so Python never asks for any area.

No extra if statement is required.

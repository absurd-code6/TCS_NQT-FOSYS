# Brute-Force/Naive Approach {O(nlogn)}

```python
def Merge(arr1,arr2):
    # Create a new list(a shallow copy) so we don't overwrite the original arr1
    combined=list(arr1)
    '''In Python, methods like .extend() and .sort() modify the list in-place. 
    This means they change the original list directly and return 
    absolutely nothing (None).'''
    combined.extend(arr2)
    combined.sort()
    return combined

# One-line Pythonic Way
'''def Merge(arr1, arr2):
    return sorted(arr1 + arr2)'''
```

```python
def merge_2_array(A,B):
    result=[]
    i=0
    j=len(B)-1
    while i<len(A) and j>=0:
        if A[i]<=B[j]:
            result.append(A[i])
            i+=1
        else:
            result.append(B[j])
            j-=1
    # If we ran out of elements in B, append the rest of A        
    while i<len(A):
        result.append(A[i])
        i+=1
    # If we ran out of elements in A, append the rest of B(moving backwards)
    while j>=0:
        result.append(B[j])
        j-=1
    return result
        
arr1=list(map(int,input().split())) #[1,3,5,7]
arr2=list(map(int,input().split())) #[10,8,6,4]
print(merge_2_array(arr1,arr2)) #[1, 3, 4, 5, 6, 7, 8, 10]
```

By utilizing the existing sorted structure with our two pointers, this code processes each element exactly once, giving it a blazing fast linear time complexity of $O(N)$

# This Problem is the Stock Buy and Sell Problem in Disguise

```python
def Rahuls_Profit(prices):
    n=len(prices)
    profit=0
    for i in range(n-1):
        for j in range(i+1,n):
            profit=max(profit,prices[j]-prices[i])
    return profit
```

This was the Brute Force Approach(TC= O(n^2))
Now Lets have a look at the approach 

with TC=O(n)

```python
def Rahuls_Profit(prices):
    # If there are fewer than 2 days, we can't buy and sell
    if len(prices) < 2:
        return 0
    min_price=prices[0]
    max_profit=0
    for p in prices:
        if p<min_price:
            min_price=p
            profit=p-min_price
        if profit>max_profit:
            max_profit=profit
    return max_profit
```

```python
def BT_Paths(root):
    if not root:
        return
    paths=[]
    def dfs(node,path):
        if node:
            path+=str(node.val)
        if not node.left and not node.right:
            paths.append(path)
        else:
            path+='->'
            dfs(node.left,path)
            dfs(node.right,path)
    dfs(root,"")
    return paths
#TC=O(n),SC=O(Ht of tree)=O(nlogn)
```

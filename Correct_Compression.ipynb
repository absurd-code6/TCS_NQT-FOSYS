```python
import re
from collections import defaultdict
def Correct_Compression(s:str)->str:
    charfreq=defaultdict(int)
    pair_matches=re.findall(r'([a-zA-Z])(\d+)',s)
    for char,freq in pair_matches:
        charfreq[char]+=int(freq)
    res=[]
    for ch in sorted(charfreq.keys()):
        res.append(f"{ch}{charfreq[ch]}")
    return "".join(res)
```

```python
s=input().strip() # "a3b2c5a1g9c2"or "b2a3a4c1b5"
print(Correct_Compression(s)) # "a4b2c7g9" or "a7b7c1"
```

```python
#What is re and what is re.findall(r'([a-zA-Z])(\d+)', s) doing? 

'''Think of re in Python as a specialized pattern-matching assistant—like 
using Ctrl + F (Find) on steroids.

Normally, Ctrl + F looks for exact text (like finding the exact word "cat"). 
The re module (short for Regular Expressions) lets you search for rules and 
patterns instead of exact words—for example, "find any single letter followed 
by one or more numbers."

Breaking Down re.findall(r'([a-zA-Z])(\d+)', s)
re.findall() goes through your string s from left to right, finds every part 
that matches your pattern rule, and returns them in a list.

Here is what each part of r'([a-zA-Z])(\d+)' means:

The r Prefix (Raw String)The r right before the quotes tells Python: 
"Treat backslashes \ as regular characters, not special escape codes." 
It ensures Python passes the pattern to re exactly as written.2. ([a-zA-Z]) -> 
Group 1: The Letter[...] (Character Set): Matches any single character 
listed inside the brackets.a-z: Any lowercase letter from a to z.
A-Z: Any uppercase letter from A to Z.
( ... ) (Capturing Group 1): Putting parentheses around [a-zA-Z] tells Python 
to isolate and save this letter separately.

(\d+) $\rightarrow$ Group 2: The Number\d: Short for "digit" (any number from 0 to 9).
+: Means "one or more" of the preceding character. 
So \d+ matches single digits like 3 or multi-digit numbers like 125.
( ... ) (Capturing Group 2): Putting parentheses around \d+ tells Python 
to isolate and save this number sequence separately.
How re.findall Combines Them Because the pattern has two set of parentheses 
(...) (...), re.findall() automatically packages every match as a pair (tuple) 
containing:(Group 1, Group 2) -> (Letter, Number)

Visual Example:

"a12b3"
   │││││
   │└┴┼┴── Matches 'a' (letter) and '12' (number) -> ('a', '12')
   └──┴─── Matches 'b' (letter) and '3'  (number) -> ('b', '3')

Python Output:

[('a', '12'), ('b', '3')]
'''
```

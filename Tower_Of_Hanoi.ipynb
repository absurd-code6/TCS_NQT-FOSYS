# Option A: The Direct Formula Method (Fastest)

Let's test the formula mathematically:

n = 1 -> 2^1 - 1 = 1

n = 2 -> 2^2 - 1 = 3

n = 3 -> 2^3 - 1 = 7

n = 4 -> 2^4 - 1 = 15

So the formula for minimum moves for n disks is simply 2^n - 1

```python
def Tower_of_Hanoi(n):
    return (2**n)-1

n=int(input("Enter the number of disks:"))
print(f"Total number of moves is {Tower_of_Hanoi(n)}") 
```

# Option B: The Recursive Method (Simulating the steps)

Thinking through small examples manually

Let's see what happens for different values of n (number of disks):

If n = 1:Just move the 1 disk directly from from to to. 

Moves = 1

If n = 2: 
Move disk 1 (smaller) to aux (1 move)

Move disk 2 (larger) to to (1 move)

Move disk 1 (smaller) from aux to to (1 move)

Moves = 1 + 1 + 1 = 3

If n = 3:
To get the largest disk (disk 3) to the target rod, we first have to move 

the top 2 disks out of the way onto the auxiliary rod.

Moving 2 disks takes 3 moves (as we saw above).

Then, move disk 3 to the target rod: 1 move.

Finally, move the 2 disks from the auxiliary rod to the target rod: another 3 moves.

Moves = 3 + 1 + 3 = 7

Spotting the Pattern

Notice the recursive pattern:

To move n disks:

Move n-1 disks out of the way: T(n-1) moves.

Move the largest single disk to the target: 1 move.

Move the n-1 disks onto the target: T(n-1) moves.

(Total Moves  T(n) = 2 x T(n-1) + 1


```python
def Tower_of_Hanoi(n):
    # Base case: if there's only 1 disk, it takes 1 move
    if n==1:
        return 1
    return 2*Tower_of_Hanoi(n-1)+1

n=int(input("Enter the number of disks:"))
print(f"Total number of moves is {Tower_of_Hanoi(n)}") 
```

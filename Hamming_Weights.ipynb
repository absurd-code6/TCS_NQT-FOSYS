```python
def Hamming_Weight(n:int)->int:
    if not n:
        return -1
    count=0
    while n>0:
        count+=n%2 #Check if rightmost bit is 1
        #OR
        if ((n&1)!=0):
            count+=1
        n//=2 #Discard or Perform the right shift
        #OR
        #n=n>>1
    return count

n=int(input())
print(n)
print(f"No of set bit representations: ",Hamming_Weight(n))
```

```python
def Brian_Kernighans_Algo(n:int)->int:
    count=0
    while n>0:
        n=n & (n-1) # Bitwise and operator in use
        count+=1
    return count
```

```python
def isSymmetric(root):
   def dfs(left,right):
    if not left and not right:
        return True #Null left & right children are symmetric technically
    if not left or not right:
        return False #If one child is Null while the oth is not then they'ren't symmetric
    return (left.val==  right.val and
    dfs(left.left,right.right) and dfs(left.right,right.left))
   return dfs(root.left,root.right)
# We need to compare the left child of one subtree with the right child of the 
#oth subtree for checking symmetricity

```

```python
class TreeNode:
    def __init__(self,val=0,left=None,right=None):
        self.val=val
        self.left=left
        self.right=right
root = TreeNode(1)
root.left = TreeNode(2, TreeNode(3), TreeNode(4))
root.right = TreeNode(2, TreeNode(4), TreeNode(3))

# Constructing Tree  1
#                   / \
#                  2   2
#                 / \ / \
#                3  4 4  3

print(isSymmetric(root))
```

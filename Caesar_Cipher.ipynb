```python
def caesar_cipher(key,msg):
    if key<0:
        print("Invalid Input")# or return "INVALID INPUT"
        return
    res=""
    for ch in msg:
        if ch.isupper():
            new_ch=chr((ord(ch)-ord('A')+key)%26 + ord('A'))
            res+=new_ch
        elif ch.islower():
            new_ch=chr((ord(ch)-ord('a')+key)%26 + ord('a'))
            res+=new_ch
        elif ch.isdigit():
            new_digit=str((int(ch)+key)%10)
            res+=new_digit
        # Space, '-' and other symbols
        else:
            res+=ch
    return res

msg=input()
key=int(input())
print(caesar_cipher(key,msg))
        
```

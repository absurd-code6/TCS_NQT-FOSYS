Take example array:[1,1,2,2,3,4,4]

XOR Approach(O(n))

```python
def Odd_occurance_xor(arr):
    single=0
    for i in arr:
        single^=i
    return single
```

Brute Force Approach(O(n))

```python
def Odd_occurance(arr):
    for i in range(0,len(arr),2):
        if arr[i]!=arr[i+1]:
         return arr[i]
# If we checked all pairs and they were all valid, 
    # the single element must be the very last element of the array.
    return arr[-1]
```

Binary Search Approach(O(logn))

```python
def Odd_Occurance_bs(arr):
    low=0
    high=len(arr)-1
    
    if low==high:
        return arr[low]
    while low<=high:
     mid=(low+high)//2
     if(arr[mid]!=arr[mid-1] and arr[mid]!=arr[mid+1]):
         return arr[mid]
     if mid%2==0:
         if(arr[mid-1]==arr[mid]):
             high=mid-1
         else:
           low=mid+1
     else:
         if arr[mid-1]==arr[mid]:
             low=mid+1
         else:
             high=mid-1
             
    return -1
        
    
    
         
```

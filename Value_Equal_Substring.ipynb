Now each group must be split into pieces of size 2 or 3.


For a group of length:

Length	Possible?	Reason

1	❌	Can't make 2 or 3

2	✅	One length-2 substring

3	✅	One length-3 substring

4	❌	Would need 2+2 (two length-2 substrings)

5	✅	2+3

6	✅	3+3

7	❌	2+2+3 (two length-2 substrings)


Also, the entire string must contain exactly one length-2 substring

```python
def value_equal_substr(s):
    i=0
    two_count=0
    while i<len(s):
        j=i
        # Find length of current group of same characters
        while j<len(s) and s[j]==s[i]:
            j+=1
        length=j-i
        # Check this group
        if length == 1 or length == 4 or length == 7:
            return False
        if length%3==2:
            # This group needs one substring of length 2
            two_count += 1
        elif length % 3 == 1:
            # Impossible (except lengths already handled)
            return False
        i=j
    return two_count==1

s=input()
if (value_equal_substr(s)):
    print('true')
else:print("false")
```
